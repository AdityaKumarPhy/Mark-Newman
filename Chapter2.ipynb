## Example 2.1  
A ball is dropped from a tower of height $h$. It has initial velocity zero and accelerates downwards under gravity. The challenge is to write a program that asks the user to enter the height in meters of the tower and a time interval $t$ in seconds, then prints on the screen the height of the ball above the ground at time $t$ after it is dropped, ignoring air resistance.

```python
import math
h = int(input("Enter the height of the tower: "))
t = int(input("Enter the time interval: "))
g = 9.81
x = (g*t**2)/2
if(h-x)<0:
    print("The ball has already reached the ground")
else:
    print("The ball is at a height of ",h-x,"meters at time ",t,"sec")
```

## Exercise 2.2: Altitude of a Satellite

A satellite is to be launched into a circular orbit around the Earth so that it orbits the planet once every $( T )$ seconds.

1. Show that the altitude $( h )$ above the Earth's surface that the satellite must have is $$\bigg[
h = \left( \frac{GMT^2}{4\pi^2} \right)^{1/3} - R
\bigg],$$ where $[G = 6.67 \times 10^{-11} \ \text{m}^3 \text{kg}^{-1} \text{s}^{-2}
]$ is Newton's gravitational constant, $[M = 5.97 \times 10^{24} \ \text{kg}]$ is the mass of the Earth, and $[R = 6371 \ \text{km}]$ is its radius.

2. Write a program that asks the user to enter the desired value of $T$ and then calculates and prints out the correct altitude in meters.

3. Use your program to calculate the altitudes of satellites that orbit the Earth:
- once a day (so-called "geosynchronous" orbit),
- once every 90 minutes,
- and once every 45 minutes.<br>
What do you conclude from the last of these calculations?

4. Technically a geosynchronous satellite is one that orbits the Earth once per **sidereal day**, which is 23.93 hours, not 24 hours. Why is this? And how much difference will it make to the altitude of the satellite?

```python
import math

# Constants
G = 6.67e-11
M = 5.97e24
R = 6371e3
T_1 = int(input("The desired value of time is :"))
# Function to calculate altitude
def altitude(T):
    return ((G * M * T**2) / (4 * math.pi**2))**(1/3) - R

# Orbit times in seconds
orbits = {
    "1 day": 24 * 60 * 60,
    "90 minutes": 90 * 60,
    "45 minutes": 45 * 60
}

# Output

print(f"2. The Altitude at time {T_1:,} sec is : {altitude(T_1):,.2f} meters \n","-"*45)
print("3.")
for name, T in orbits.items():
    h = altitude(T)
    print(f"{name:12} -> {h:,.2f} meters") 
print("Conclusion \n A satellite cannot orbit the Earth with a period as short as 45 minutes because \n the required orbital radius would be smaller than Earth’s radius itself.")
print("-"*45,"\n 4.")

t_sidereal = 23.93*60*60
t_real = 24*60*60
h_1 = altitude(t_sidereal)
h_2 = altitude(t_real)
print(f"The Sidereal Altitude -> {h_1:,.2f} meters  \n The Real Altitude -> {h_2:,.2f} meters")
print(f"The difference it will make to the altitude of the satelite is -> {h_2-h_1:,.2f}meters")
```

## Example 2.2: Converting Polar Coordinates

Suppose the position of a point in two-dimensional space is given to us in polar coordinates $r, \theta$ and we want to convert it to Cartesian coordinates $x, y$. How would we write a program to do this? The appropriate steps are:

1. Get the user to enter the values of $r$ and $\theta$.

2. Convert those values to Cartesian coordinates using the standard formulas:

$$
x = r \cos \theta, \qquad y = r \sin \theta
$$

3. Print out the results.

```python
from math import sin, cos, pi
from tabulate import tabulate #module for table outputs

r = float(input("Enter r : "))
d = float(input("Enter θ in degree : "))

theta = d * pi / 180

# Polar → Cartesian
x = r * cos(theta)
y = r * sin(theta)

# Table
table = [
    [f"({r:.2f}, {theta:.2f})",
     f"({x:.2f}, {y:.2f})"]
]

print(tabulate(table,
               headers=["Polar Coordinates", "Cartesian Coordinates"],
               tablefmt="grid"))

print(f"\nThe value of ({r:.2f}, {theta:.2f}) in Cartesian coordinates is ({x:.2f}, {y:.2f})")
```

$$

$$
m = 9.11 \times 10^{-31}\ \text{kg}
$$

and energy

$$
E = 10\ \text{eV}
$$

encountering a potential step of height

$$
V = 9\ \text{eV}
