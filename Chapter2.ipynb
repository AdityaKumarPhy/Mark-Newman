### Example 2.1  
A ball is dropped from a tower of height $h$. It has initial velocity zero and accelerates downwards under gravity. The challenge is to write a program that asks the user to enter the height in meters of the tower and a time interval $t$ in seconds, then prints on the screen the height of the ball above the ground at time $t$ after it is dropped, ignoring air resistance.

```python
import math
h = int(input("Enter the height of the tower: "))
t = int(input("Enter the time interval: "))
g = 9.81
x = (g*t**2)/2
if(h-x)<0:
    print("The ball has already reached the ground")
else:
    print("The ball is at a height of ",h-x,"meters at time ",t,"sec")
```

### Exercise 2.2: Altitude of a Satellite

A satellite is to be launched into a circular orbit around the Earth so that it orbits the planet once every $( T )$ seconds.

1. Show that the altitude $( h )$ above the Earth's surface that the satellite must have is $$\bigg[
h = \left( \frac{GMT^2}{4\pi^2} \right)^{1/3} - R
\bigg],$$ where $[G = 6.67 \times 10^{-11} \ \text{m}^3 \text{kg}^{-1} \text{s}^{-2}
]$ is Newton's gravitational constant, $[M = 5.97 \times 10^{24} \ \text{kg}]$ is the mass of the Earth, and $[R = 6371 \ \text{km}]$ is its radius.

2. Write a program that asks the user to enter the desired value of $T$ and then calculates and prints out the correct altitude in meters.

3. Use your program to calculate the altitudes of satellites that orbit the Earth:
- once a day (so-called "geosynchronous" orbit),
- once every 90 minutes,
- and once every 45 minutes.<br>
What do you conclude from the last of these calculations?

4. Technically a geosynchronous satellite is one that orbits the Earth once per **sidereal day**, which is 23.93 hours, not 24 hours. Why is this? And how much difference will it make to the altitude of the satellite?

```python
import math

# Constants
G = 6.67e-11
M = 5.97e24
R = 6371e3
T_1 = int(input("The desired value of time is :"))
# Function to calculate altitude
def altitude(T):
    return ((G * M * T**2) / (4 * math.pi**2))**(1/3) - R

# Orbit times in seconds
orbits = {
    "1 day": 24 * 60 * 60,
    "90 minutes": 90 * 60,
    "45 minutes": 45 * 60
}

# Output

print(f"2. The Altitude at time {T_1:,} sec is : {altitude(T_1):,.2f} meters \n","-"*45)
print("3.")
for name, T in orbits.items():
    h = altitude(T)
    print(f"{name:12} -> {h:,.2f} meters") 
print("Conclusion \n A satellite cannot orbit the Earth with a period as short as 45 minutes because \n the required orbital radius would be smaller than Earth’s radius itself.")
print("-"*45,"\n 4.")

t_sidereal = 23.93*60*60
t_real = 24*60*60
h_1 = altitude(t_sidereal)
h_2 = altitude(t_real)
print(f"The Sidereal Altitude -> {h_1:,.2f} meters  \n The Real Altitude -> {h_2:,.2f} meters")
print(f"The difference it will make to the altitude of the satelite is -> {h_2-h_1:,.2f}meters")
```

### Example 2.2: Converting Polar Coordinates

Suppose the position of a point in two-dimensional space is given to us in polar coordinates $r, \theta$ and we want to convert it to Cartesian coordinates $x, y$. How would we write a program to do this? The appropriate steps are:

1. Get the user to enter the values of $r$ and $\theta$.

2. Convert those values to Cartesian coordinates using the standard formulas:

$$
x = r \cos \theta, \qquad y = r \sin \theta
$$

3. Print out the results.

```python
from math import sin, cos, pi
from tabulate import tabulate #module for table outputs

r = float(input("Enter r : "))
d = float(input("Enter θ in degree : "))

theta = d * pi / 180

# Polar → Cartesian
x = r * cos(theta)
y = r * sin(theta)

# Table
table = [
    [f"({r:.2f}, {theta:.2f})",
     f"({x:.2f}, {y:.2f})"]
]

print(tabulate(table,
               headers=["Polar Coordinates", "Cartesian Coordinates"],
               tablefmt="grid"))

print(f"\nThe value of ({r:.2f}, {theta:.2f}) in Cartesian coordinates is ({x:.2f}, {y:.2f})")
```

### Exercise 2.4

A spaceship travels from Earth in a straight line at relativistic speed $v$ to another planet $x$ light years away.

Write a program to ask the user for the value of $x$ and the speed $v$ as a fraction of the speed of light $c$, then print out the time in years that the spaceship takes to reach its destination:

1. In the rest frame of an observer on Earth
2. As perceived by a passenger on board the ship

Use your program to calculate the answers for a planet $10$ light years away with $v = 0.99c$.

---

#### <span style="color: green;">Analytical Solution </span>
1. In the rest frame of Earth's Observer 
    - Basic Relation $speed = \frac{dist}{time}$
    - Speed as a fraction of Speed of Light, $\boxed{v=\beta c}$  where $$\beta = \frac{v}{c}$$
    - 1$c$=1 $lightyear$
    - So, the time will become $$t_{earth} = \frac{x}{\beta} $$ , this will give time in years 

2. Time Measured by Passenger 
    - now here we will use time dilation $$t_{passenger} = t_{earth} \sqrt{1-{\beta}^2}$$
    - using the Lorentz factor $$\gamma = \frac{1}{\sqrt{1-{\beta}^2}}$$
    - therefore $$t_{passenger}=\frac{t_{earth}}{\gamma}$$



```python
import math
from tabulate import tabulate

def gamma(v):
    return 1 / math.sqrt(1 - v**2)

def EarObs(x, v):
    return x / v

def PassObs(x, v):
    return EarObs(x, v) / gamma(v)


# User input
x = float(input("Enter distance (light years): "))
v = float(input("Enter speed as fraction of c: "))

# Table storage
table = []


#  Demo dataset from question
demo_x = 10
demo_v = 0.99

table.append([
    demo_x,
    demo_v,
    round(EarObs(demo_x, demo_v), 3),
    round(PassObs(demo_x, demo_v), 3)
])


# User dataset
table.append([
    x,
    v,
    round(EarObs(x, v), 3),
    round(PassObs(x, v), 3)
])


# Iterative datasets
for i in range(2):

    x += 10
    v += 0.01

    # prevent impossible speed
    if v >= 1:
        break

    table.append([
        x,
        round(v, 2),
        round(EarObs(x, v), 3),
        round(PassObs(x, v), 3)
    ])


# Headers
headers = [
    "Distance (ly)",
    "Speed (c)",
    "Earth Observer (years)",
    "Passenger (years)"
]


# Print table
print(tabulate(table, headers=headers, tablefmt="fancy_grid"))
```

### Exercise 2.5: Quantum Potential Step

A well-known quantum mechanics problem involves a particle of mass $m$ that encounters a one-dimensional potential step.



![Quantum Potential Step](Assets/tunnel.png)

The particle with initial kinetic energy $E$ and wavevector

$$
k_1 = \frac{\sqrt{2mE}}{\hbar}
$$

enters from the left and encounters a sudden jump in potential energy of height $V$ at position $x = 0$.

By solving the Schrödinger equation, one can show that when $E > V$ the particle may either:

1. Pass the step, in which case it has a lower kinetic energy of $E - V$ on the other side and a correspondingly smaller wavevector

$$
k_2 = \frac{\sqrt{2m(E - V)}}{\hbar}
$$

2. Be reflected, keeping all of its kinetic energy and an unchanged wavevector but moving in the opposite direction.

The probabilities $T$ and $R$ for transmission and reflection are given by

$$
T = \frac{4k_1k_2}{(k_1 + k_2)^2}
$$

and

$$
R = \left(\frac{k_1 - k_2}{k_1 + k_2}\right)^2
$$

Suppose we have a particle with mass equal to the electron mass

$$
m = 9.11 \times 10^{-31}\ \text{kg}
$$

and energy

$$
E = 10\ \text{eV}
$$

encountering a potential step of height

$$
V = 9\ \text{eV}
$$

Write a Python program to compute and print out the transmission and reflection probabilities using the formulas above.

```python
import numpy as np

m = 9.11*10**-31
eV = 1.602e-19  
h_bar = 1.054e-34     #reduced planks const. (j/s)

#Energies in Joules
E = 10*eV    #in eV
V = 9*eV     #eV

# lets define the wave vectors k_1 and k_2
k_1 = np.sqrt(2*m*E)/h_bar
k_2 = np.sqrt(2*m*(E-V))/h_bar

# lets compute the values of Transmission and Reflection probability 

T = (4*k_1*k_2)/(k_1+k_2)**2
R = ((k_1-k_2)/(k_1+k_2))**2

# Now lets print the values of the T and R

print(f"The Transmission Probability is : {T}")
print(f"Reflection Probability is : {R}")
print(f"Since the Probability is conserved so T + R = {T+R}")
```
